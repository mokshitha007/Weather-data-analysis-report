```python
# import libraries
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#Weather Dataset from google
data = {
    "Date": [
        "2026-02-01","2026-02-02","2026-02-03","2026-02-04",
        "2026-02-05","2026-02-06","2026-02-07","2026-02-08",
        "2026-02-09","2026-02-10","2026-02-11","2026-02-12",
        "2026-02-13","2026-02-14","2026-02-15","2026-02-16",
        "2026-02-17"
    ],

    "MinTemp": [22.0,22.2,20.4,21.9,21.0,21.8,20.9,21.4,20.2,20.8,20.4,20.5,21.9,22.3,21.7,21.2,20.7],

    "AvgTemp": [25.9,25.7,25.0,26.0,25.4,25.7,25.5,25.3,25.2,25.2,24.4,25.0,26.1,26.6,26.1,25.7,25.3],

    "MaxTemp": [30.5,29.9,30.7,31.2,31.1,30.8,31.2,31.4,31.1,30.7,29.0,30.9,31.9,31.7,32.1,31.5,30.4],

    "Humidity": [69,69,71,71,71,69,69,71,71,69,69,69,69,69,69,69,69]
}

df = pd.DataFrame(data)

df
```

```python
import pandas as pd
```

```python
df['Date'] = pd.to_datetime(df['Date'])

df.set_index('Date', inplace=True)

df
```

```python
plt.figure(figsize=(10,5))

plt.plot(df['AvgTemp'])

plt.title("Average Temperature - Feb 2026 (Chennai)")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")

plt.show()
```

```python
df['Day'] = np.arange(len(df))

X = df[['Day']]
y = df['AvgTemp']
```

```python
import pandas as pd

data = {
    "Date": ["2026-02-01", "2026-02-02"],
    "AvgTemp": [25.9, 25.7]
}

df = pd.DataFrame(data)

print(df)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    shuffle=False
)
```

```python
model = LinearRegression()

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MSE:", mse)
print("R2 Score:", r2)
```

```python
plt.figure(figsize=(10,5))

plt.plot(y_test.values, label="Actual")
plt.plot(y_pred, label="Predicted")

plt.title("Actual vs Predicted Temperature")
plt.legend()

plt.show()
```

```python
future_days = np.array([[len(df)+i] for i in range(10)])

future_temp = model.predict(future_days)

future_temp
```
